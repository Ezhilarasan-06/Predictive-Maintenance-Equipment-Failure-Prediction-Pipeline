```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
silver_df = spark.table("predictive_cataglog.silver_layer.silver_table")
```

```python
silver_df.limit(5).display()
```

```python

failure_df = silver_df.groupBy("engine_id") \
    .agg(max("cycle").alias("failure_cycle"))
```

```python
labeled_df = silver_df.join(
    failure_df,
    on="engine_id",
    how="inner"
)
```

```python
labeled_df.limit(5).display()
```

```python
labeled_df = labeled_df.withColumn(
    "RUL",
    col("failure_cycle") - col("cycle")
)
```

```python
Threshhold_point=30

labeled_df = labeled_df.withColumn(
    "failure_label",
    when(col("RUL") <= Threshhold_point, 1).otherwise(0)
)
```

```python
labeled_df = labeled_df.withColumn(
    "RUL_capped",
    least(col("RUL"), lit(125))
)
```

```python

labeled_df.select([count(when(col(c).isNull(), c)).alias(c) for c in labeled_df.columns]).display()
```

```python
final_labeled_df=labeled_df.select(
    "engine_id",
    "cycle",
    "RUL",
    "RUL_capped",
    "failure_label",
    *[c for c in labeled_df.columns if "roll_mean" in c or "delta" in c]
)
```

```python
final_labeled_df.limit(5).display()
```

```python
%sql
create database if not exists predictive_cataglog.gold_layer
```

```python
final_labeled_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("predictive_cataglog.gold_layer.labeled_table")
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
pred_df = spark.table("predictive_cataglog.gold_layer.risk_score_table")
label_df = spark.table("predictive_cataglog.gold_layer.labeled_table")
```

```python
pred_vs_actual_df = pred_df.join(
    label_df,
    on=["engine_id", "cycle"],
    how="inner"
)
```

```python
pred_vs_actual_df = pred_vs_actual_df.select(
    "engine_id",
    "cycle",
    "prediction",
    "failure_label",
    "failure_probability",
    "risk_level"
)
```

```python
pred_vs_actual_df.limit(5).display()
```

```python
pred_vs_actual_df.write.\
    mode("overwrite").\
    saveAsTable("predictive_cataglog.gold_layer.feature_table"
)
```

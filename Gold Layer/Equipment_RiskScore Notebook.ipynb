```python
%sql
SELECT 
  COUNT(DISTINCT engine_id) AS total_engines
FROM predictive_cataglog.gold_layer.risk_score_table;
```

```python
%sql
SELECT 
  risk_level,
  COUNT(DISTINCT engine_id) AS engine_count
FROM predictive_cataglog.gold_layer.risk_score_table
GROUP BY risk_level;
```

```python
%sql
SELECT 
  COUNT(DISTINCT engine_id) AS high_risk_engines
FROM predictive_cataglog.gold_layer.risk_score_table
WHERE risk_level = 'HIGH';
```

```python
%sql
SELECT 
  engine_id,
  MAX(failure_probability) AS max_failure_probability,
  risk_level
FROM predictive_cataglog.gold_layer.risk_score_table
GROUP BY engine_id, risk_level
ORDER BY max_failure_probability DESC
LIMIT 10;
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
silver_df = spark.table("predictive_cataglog.silver_layer.silver_table")
```

```python
feature_cols = [c for c in silver_df.columns if "roll" in c or "delta" in c]

null_features = {
    c: silver_df.filter(col(c).isNull()).count()
    for c in feature_cols
}
null_features

```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag

w = Window.partitionBy("engine_id").orderBy("cycle")

silver_df.withColumn(
    "cycle_gap",
    col("cycle") - lag("cycle").over(w)
).filter(col("cycle_gap") > 1).count()
```

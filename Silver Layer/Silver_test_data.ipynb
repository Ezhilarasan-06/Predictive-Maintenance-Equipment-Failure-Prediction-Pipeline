```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
test_silver_df = spark.table("predictive_cataglog.bronze_layer.test_raw_table")
```

```python
test_silver_df.limit(10).display()
```

```python
w = Window.partitionBy("engine_id").orderBy("cycle")

for i in range(1, 22):
    test_silver_df = test_silver_df.withColumn(
        f"s{i}_roll_mean", avg(col(f"s{i}")).over(w.rowsBetween(-5, 0))
    ).withColumn(
        f"s{i}_delta", col(f"s{i}") - lag(col(f"s{i}"), 1).over(w)
    )
```

```python
test_silver_feature_df = test_silver_df.dropna()
```

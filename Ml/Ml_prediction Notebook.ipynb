```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
test_raw_df = (
    spark.read
    .option("sep", " ")
    .option("inferSchema", True)
    .csv("/Volumes/predictive_cataglog/default/datasets/test_FD001.txt")
)
```

```python
null_columns = [c for c in test_raw_df.columns if test_raw_df.filter(col(c).isNotNull()).count() == 0]
test_raw_df = test_raw_df.drop(*null_columns)
```

```python
test_raw_df.limit(5).display()
```

```python
cols = ["engine_id", "cycle", "op1", "op2", "op3"] + \
       [f"s{i}" for i in range(1, 22)]

test_raw_df = test_raw_df.toDF(*cols)

```

```python
test_raw_df.limit(5).display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import avg, col, lag

w = Window.partitionBy("engine_id").orderBy("cycle")

for i in range(1, 22):
    test_raw_df = test_raw_df.withColumn(
        f"s{i}_roll_mean", avg(col(f"s{i}")).over(w.rowsBetween(-5, 0))
    ).withColumn(
        f"s{i}_delta", col(f"s{i}") - lag(col(f"s{i}"), 1).over(w)
    )
```

```python
test_silver_df = test_raw_df.dropna()
```

```python
test_silver_df.limit(5).display()
```

```python
from pyspark.ml import PipelineModel

model = PipelineModel.load(
    "/Volumes/predictive_cataglog/default/my_model/cmapss_lr_model/"
)
```

```python
test_predictions = model.transform(test_silver_df)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import DoubleType

get_prob = udf(lambda v: float(v[1]), DoubleType())

failure_prob_df = test_predictions.select(
    "engine_id",
    "cycle",
    get_prob("probability").alias("failure_probability")
)
```

```python
failure_prob_df.limit(5).display()
```

```python
from pyspark.sql.functions import when

risk_score_df = failure_prob_df.withColumn(
    "risk_level",
    when(col("failure_probability") >= 0.6, "HIGH")
    .when(col("failure_probability") >= 0.3, "MEDIUM")
    .otherwise("LOW")
)
```

```python
risk_score_df.limit(5).display()
```

```python
risk_score_df.write.\
    format("delta").\
    mode("overwrite").\
    saveAsTable("predictive_cataglog.gold_layer.Risk_Score_Table")
```

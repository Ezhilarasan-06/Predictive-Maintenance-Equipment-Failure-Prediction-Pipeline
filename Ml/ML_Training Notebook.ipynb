```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
labeled_df= spark.read.table("predictive_cataglog.gold_layer.labeled_table")
```

```python
labeled_df.limit(5).display()
```

```python
label_col = "failure_label"

feature_cols = [
    c for c in labeled_df.columns
    if c.startswith("s") and ("roll_mean" in c or "delta" in c)
]
```

```python
ml_df = labeled_df.select(
    "engine_id", "cycle", label_col, *feature_cols
).dropna()
```

```python
ml_df.limit(5).display()
```

```python
engine_ids = ml_df.select("engine_id").distinct()

train_engines, val_engines = engine_ids.randomSplit([0.8, 0.2], seed=42)

train_df = ml_df.join(train_engines, "engine_id")
val_df   = ml_df.join(val_engines, "engine_id")
```

```python
from pyspark.ml.feature import VectorAssembler

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="features",
    handleInvalid="skip"
)
```

```python
from pyspark.ml.classification import LogisticRegression

lr = LogisticRegression(
    featuresCol="features",
    labelCol=label_col,
    probabilityCol="probability",
    predictionCol="prediction",
    maxIter=50
)
```

```python
from pyspark.ml import Pipeline

pipeline = Pipeline(stages=[
    assembler,
    lr
])
```

```python
model = pipeline.fit(train_df)

```

```python
predictions = model.transform(val_df)
```

```python
predictions.groupBy("prediction").count().show()
```

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

evaluator = BinaryClassificationEvaluator(
    labelCol=label_col,
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

auc = evaluator.evaluate(predictions)
print("Validation AUC:", auc)
```

```python
model.write() \
    .overwrite() \
    .save("/Volumes/predictive_cataglog/default/my_model/cmapss_lr_model")
```
